## 1. Ziel und Rangordnung ✅ (26.11.2026)

**Gesucht sind alle markierten, öffentlichen Parkplätze** im Gebiet – egal welche Art (Standard, Blaue Zone, Güterumschlag, Taxi, Elektro …).

**Grundsatz:** Wir setzen uns kein Genauigkeitsziel. Wir machen das Beste, was die Daten hergeben, **ohne Verfälschungen**: Nichts wird geschätzt, was nicht belegt ist, und jede Ergänzung wird getrennt ausgewiesen.

| Stufe | Frage | Stand |
| --- | --- | --- |
| **0. Punkt** | Ist hier ein **aufgemaltes Feld**? (Fläche erkannt, alles Weitere noch unbekannt) Auf dieser Stufe wird jedes aufgemalte Einzelfeld gefunden, auch ein gelbes Feld mit Kreuz ✅ | dieser Pilot |
| **1. Form und Parkart** | Was für ein Feld ist es? Parkart `LPU` längs, `QPU` quer, `SCHPU` schräg (bestimmt, wie viel Fläche ein Platz braucht) und Form: gewöhnliches Feld oder gelbes Feld mit Kreuz | dieser Pilot ✅ |
| 2. Farbe | Welche Farbe hat die Markierung (weiss, blau, gelb/orange, grün)? | später |
| 3. Kategorie | Was bedeutet die Farbe (weiss → Standard, blau → Blaue Zone, grün → Elektro …)? | später |

**Erst finden, dann einordnen ✅ (26.11.2026).** Das Netz muss auf Stufe 0 nur eines lernen: Hier ist ein Feld aufgemalt. Was für ein Feld es ist, kommt Stufe um Stufe dazu. Das ist einfacher und ehrlicher, als schon beim Finden zu entscheiden, was zählt.

> **Stand heute (27.11.2026).** Stufe 1 ist in diesem Durchgang vereinfacht: statt drei Parkarten zwei Klassen, *eng* (quer oder schräg) und *weit* (längs), aus dem Abstand der Ringe (Abschnitt 11, E1). Die Form (`✕`, `?`) ist erfasst und wird getrennt ausgewiesen (11.2, N1).


Beispiel gelbes Feld mit Kreuz (Sichtprüfung, Felder Nr. 9 und Nr. 20): Nach der Signalisationsverordnung des Bundes (SSV, Art. 79a) ist das ein **Parkverbotsfeld**. Auf Stufe 0 wird es als aufgemaltes Feld gefunden und mitgezählt – so führt es auch die DAV, der Vergleich bleibt also gleich gegen gleich. Ab Stufe 1 ist seine Form bekannt; dann wird es **getrennt ausgewiesen** und kann aus der Zahl der eigentlichen Parkplätze herausgerechnet werden. In der eigenen Ground Truth bekommt jedes Feld dafür das Merkmal `form`.

**Drei Rollen, sauber getrennt ✅ (26.11.2026):**

| Rolle | Quelle | Wozu |
| --- | --- | --- |
| **Lernen** (Lösungsblatt fürs Netz) | **eigene Ground Truth** im Lerngebiet: Felder von Hand auf dem Luftbild gesetzt | sicher, weil direkt auf dem Bild, das gelernt wird – keine Versetzung, keine veralteten Felder |
| **Prüfen** | **eigene Ground Truth in den Testzonen** (beim Lernen nie gesehen) | zeigt, wie gut das Modell gegenüber einer sicheren Lösung ist |
| **Vergleich am Schluss** | amtliche Daten (DAV) desselben Jahres | **Delta** = Gesamtzahl erkannter Parkplätze − Gesamtzahl laut DAV |

Beim Delta ist keine der beiden Zahlen sicher die Wahrheit: Das Modell kann sich irren, die amtlichen Daten können veraltet oder lückenhaft sein. Das Delta zeigt nur, **wo und wie stark** sie sich unterscheiden. Erst die Prüfung in den Testzonen trennt die beiden Fehlerquellen (formal in Abschnitt 7.9).

> **Stand heute (27.11.2026).** Die Testzonen heissen in diesem Durchgang Testblöcke (Abschnitt 11, E3). Für Versuch A sind die amtlichen Daten die Strassenparkplätze der Stadt mit Stand 31.12.2021 (E11). Das Delta wird mit der Ground Truth in drei Teile zerlegt (12.9).


**Bezeichnungen:** Es gelten die Original-Kürzel der DAV (`orientierung`, `art`).

**Platzgrössen aus amtlichen Vorgaben ✅ (26.11.2026).** Wir messen nicht selbst, wie gross ein Parkfeld im Durchschnitt ist, sondern übernehmen die **offiziellen Masse** der Schweizer Norm VSS 40 291 (*Parkieren; Anordnung und Geometrie der Parkierungsanlagen für Personenwagen und Motorräder*, Ausgabe 2021-12). Die Norm selbst ist kostenpflichtig; ihre Masse stehen aber in **frei zugänglichen amtlichen Unterlagen**, die sich auf sie berufen (Ablage: Ordner `quellen/`, Liste in `quellen_liste.py`):

| Parkart | Mass laut amtlicher Unterlage | Quelle |
| --- | --- | --- |
| `LPU` längs | Breite **1,90 m**; Länge **5,00 m** am Anfang und Ende einer Reihe, **6,00 m** (bis 6,50 m) für Felder in der Mitte (Platz zum Einparken) | Kanton Bern, Arbeitshilfe Parkierung 2026, Kap. 5.3; Stadt Bern, Normalie 2.10.22; Stadt Zürich DAV, Normblatt Längsparkfeld 2022 (Breite 1,90 m) |
| `QPU` quer | Tiefe **5,00 m**; Breite **2,50 m** bis 2,80 m, je nach Breite der Fahrgasse davor (2,50 m bei 6,50 m Fahrgasse) | Kanton Bern, Kap. 5.3; Stadt Zürich DAV, Normblatt Senkrechtparkfeld 2022 (Tiefe 5,00 m) |
| `SCHPU` schräg | Feldbreite **2,50 m** (quer zum Auto gemessen); bei 45°: 3,55 m entlang der Strasse, Tiefe 4,80–4,90 m; bei 60°: 2,90 m entlang der Strasse, Tiefe 5,25 m | Stadt Bern, Normalie 2.10.22; Stadt Zürich DAV, Normblatt Schrägparkfeld 2024 (45°, Tiefe 4,80 m) |
| `Invalid` (Gehbehinderte) | längs **8,00 × 1,90 m** (plus 1,40 m Ein-/Aussteigezone); quer **5,00 × 3,50 m**; gelb markiert | Stadt Bern, Normalie 2.10.22 |
| `Elektro` | gewöhnliche Masse; gelb markiert, Symbol "Ladestation" | Kanton Bern, Kap. 6.4.2 |

**Was dabei auffällt** (am 26.11.2026 besprochen; der Entscheid folgt direkt darunter).
1. Die Norm gibt **Mindestwerte und Bereiche**, keine einzelne Zahl (z. B. `LPU` 5,00 oder 6,00 m, `QPU` 2,50 bis 2,80 m). Sie beruht auf einem Bemessungsfahrzeug von 1,85 × 4,90 m.
2. Die Norm gilt für **neue** Anlagen (Ausgabe 2021). Viele Felder in der Zürcher Altstadt sind älter. Vorwissen aus Pilot 1 (dort aus den Abständen der DAV-Punkte gemessen): `LPU` im Mittel 5,83 m (passt zwischen 5,00 und 6,00 m), `SCHPU` rund 59° (passt zu 60°), aber `QPU` nur **2,33 m** statt mindestens 2,50 m.
3. **Folge für die Zählung:** Anzahl = Fläche ÷ Platzfläche. Sind die echten Querfelder 2,33 m breit und wir teilen durch eine Platzfläche mit 2,50 m, zählen wir bei Querfeldern rund 7 % zu wenig (2,33 ÷ 2,50 = 0,93) – auch wenn die Fläche perfekt erkannt ist. Formal: Die Platzfläche $a_k$ muss zur Fläche passen, die ein Feld im Lösungsblatt wirklich einnimmt (Abschnitt 7.7).

**Entscheid ✅ (26.11.2026): Norm plus eigene Ground Truth.**

| Mass | Quelle |
| --- | --- |
| **Tiefe** des Feldes quer zur Strasse (`LPU` 1,90 m; `QPU` 5,00 m; `SCHPU` je nach Winkel 4,80–5,25 m) | Norm (lässt sich aus Punkten nicht messen) |
| **Abstand von Feld zu Feld entlang der Strasse** | eigene Ground Truth im Lerngebiet (mittlerer Abstand benachbarter Felder je Parkart); die Norm dient als Kontrolle |

Daraus folgt **eine** Formel für alle Parkarten: Platzfläche = Abstand entlang der Strasse × Tiefe quer zur Strasse. Die Kästchen im Lösungsblatt werden genauso gezeichnet (so breit wie der Abstand zum Nachbarn), damit ein Feld im Lösungsblatt genau eine Platzfläche einnimmt. Nachgerechnet in `formeln_pruefen`, Prüfung 12. Für einzelne Felder ohne Nachbarn gilt das Normmass.

**Vom Punkt zum Kästchen.** Erkannt wird zuerst ein Ort (Stufe 0). Dort setzen wir einen Punkt und zeichnen um ihn das Kästchen in Normgrösse der Parkart (Stufe 1).

> **Stand heute (27.11.2026).** Umgesetzt in den Notebooks 03 und 04 (Abschnitt 11, E5): Die Breite des Kästchens ist der typische Abstand in der Reihe, gemessen in den Lernblöcken; die Tiefe kommt aus der Norm. Einzeln stehende Felder bekommen **kein** Kästchen nach Normmass: Ihre Richtung ist unbekannt, sie werden nicht bewertet und getrennt ausgewiesen (E4).

## 4. Verdeckung: Parkfelder unter Bäumen und im Schatten ✅ (26.11.2026)

> **Stand heute (27.11.2026).** Der Lückenschluss (4.1 bis 4.9) ist ein durchdachtes Konzept, wird in diesem Durchgang aber **nicht angewendet** (4.6, 8.2 W4, Abschnitt 11, E12). ### 4.1 Zwei Arten zu ergänzen

| | Statistische Schätzung (Hellekes 2022) | **Lückenschluss mit Beweis** |
| --- | --- | --- |
| Frage | Wie viele Plätze hat eine Strasse dieser Art *typischerweise*? | Hier ist eine Reihe erkannter Felder unterbrochen – geht sie unter dem Baum weiter? |
| Setzt Felder, wo nichts zu sehen ist? | ja | nein – nur in nachweislich verdeckten Lücken |
| Stand | ✅ **weggelassen** (26.11.2026): würde Plätze setzen, nur weil man sie erwartet | ✅ **gewählte Richtung** (26.11.2026), Einzelheiten 💬 |

### 4.2 Zeichen für die Gedankenspiele

Alles liegt an derselben Strasse, auf derselben Strassenseite.

| Zeichen | Bedeutung |
| --- | --- |
| `P` | Parkfeld **erkannt** (im Bild sichtbar) |
| `B` | **verdeckt** – unter einer Baumkrone (im Infrarot nachgewiesen). Schatten allein gilt **nicht** als verdeckt ✅ (Abschnitt 8, W5) |
| `A` | **sichtbarer Asphalt ohne Parkfeld** – man sieht, dass dort keines ist |
| `t` | Parkfeld **nur teilweise sichtbar** (z. B. ein Drittel) |
| `p` | Parkfeld **ergänzt** |
| `?` | unklar |

### 4.3 Mittelstück und Randstück

```
Mittelstück:   P  P  B  B  B  P  P     links UND rechts erkannte Felder
                     p  p  p           → die Reihe geht sehr wahrscheinlich durch

Randstück:     B  B  B  P  P  P  P     nur auf EINER Seite erkannte Felder
               ?  ?  ?                 → beginnt die Reihe unter den Bäumen
                                         oder erst beim ersten sichtbaren Feld?
```

Beim Mittelstück ist die Reihe von beiden Seiten "eingeklemmt". Beim Randstück fehlt der zweite Beleg: Reihen beginnen und enden oft an Einfahrten oder Kreuzungen.

### 4.4 Die Fälle

**Fall 1 – Verdeckung am Anfang (Randstück)**
```
B  B  B  P  P  P  P
?  ?  ?
```
Nur rechts erkannte Felder. **Entscheid ✅ (26.11.2026): gezählt werden nur die 4 erkannten Felder.** Unter den Bäumen wird nichts ergänzt und keine Spanne ausgewiesen. Grund: Reihen beginnen oft an Einfahrten, Kreuzungen oder knicken ab (z. B. L-förmige Anordnung um eine Ecke); ohne Beleg auf der anderen Seite wäre jede Ergänzung eine Vermutung.

**Fall 2 – Verdeckung zwischen erkannten Feldern (Mittelstück)**
```
P  B  B  B  B  B  P
   p  p  p  p  p
```
Der klare Fall: **ergänzen**, Anzahl = Länge der Lücke ÷ Platzlänge. Aber: Je länger die Lücke, desto eher versteckt sich darunter eine Einfahrt. Darum eine **Höchstlänge**; ist die Lücke länger, wird nicht ergänzt.

**Entscheid ✅ (26.11.2026): Die Höchstlänge wird aus der eigenen Ground Truth gemessen, nicht gesetzt.** Im Lerngebiet prüfen wir alle verdeckten Lücken: Bis zu welcher Länge enthalten sie wirklich durchgehend Felder? Die Grenze liegt bei der grössten Länge, bei der das noch in mindestens **95 von 100** Fällen stimmt ✅ (26.11.2026).

**Woher kommt "95 von 100"?** Nicht aus den Quellen: Der Lückenschluss ist unser eigener Ansatz, die Arbeiten von Henry und Hellekes geben dafür keinen Wert vor. Die Zahl ist eine **Abwägung**: Wie viel schlimmer ist ein erfundener Parkplatz als ein übersehener?

| verlangte Sicherheit | bedeutet: ein erfundener Platz wiegt so schwer wie … übersehene | nötig, um sie zu belegen: mindestens … geprüfte Lücken ohne einen einzigen Fehler |
| --- | --- | --- |
| 90 von 100 | 9 | 29 |
| **95 von 100** | **19** | **59** |
| 99 von 100 | 99 | 299 |

Die rechte Spalte ist entscheidend: Eine Sicherheit kann man nur behaupten, wenn man genug Lücken geprüft hat. Ob Kreis 1 überhaupt 59 verdeckte Lücken hat, wissen wir erst mit der Ground Truth. Hat er weniger, können wir 95 von 100 nicht belegen – dann wird **nicht ergänzt** ✅. Das ist kein Scheitern, sondern das ehrliche Ergebnis. Die Rechnung dazu steht in 7.8. Solange die Ground Truth fehlt, wird in vollständig verdeckten Lücken **gar nicht** ergänzt.

**Fall 3 – Mischung**
```
B  P  P  P  B  B  B  B  P  P  P  B
?           p  p  p  p           ?
```
Die mittlere Lücke wird ergänzt (Fall 2). Anfang und Ende sind Randstücke (Fall 1): Dort wird nichts ergänzt.

**Fall 4 – sichtbarer Asphalt unter den Bäumen**
```
P  B  B  B  B  B  A  B  B  B  P  P  P
   ?  ?  ?  ?  ?  ✗  ?  ?  ?
```
Das Stück `A` **beweist**, dass die Reihe dort unterbrochen ist. Es entstehen zwei Randstücke (links von `A`, rechts von `A`), und in Randstücken wird nichts ergänzt. **Über sichtbaren Asphalt hinweg wird nie ergänzt.**

**Fall 5 – teilweise sichtbare Felder mit Muster**
```
P  P  t  t  t  P        t = nur ein Drittel sichtbar (z. B. Ende der Markierung ragt unter der Krone hervor)
      p  p  p
```
Hier gibt es mehr als eine blosse Lücke: Man sieht **Stücke von Markierungen in regelmässigem Abstand**. Das Muster setzt sich fort – ein stärkerer Beleg als in Fall 2.

**Entscheid ✅ (26.11.2026): Mindestens ein Drittel des Feldes muss sichtbar sein**, und das Muster muss sich wiederholen (mehrere Felder nebeneinander). Der Abstand der Querstriche muss zur Normbreite der Parkart passen.

**Beispiel: Querparkfelder, deren hinterer Teil unter einem Dach liegt** (Draufsicht)
```
   ████████████████████████████████████   <- Dach: hintere zwei Drittel verdeckt
   ████████████████████████████████████
   ─────┬─────┬─────┬─────┬─────┬─────     <- Dachkante
        │     │     │     │     │          <- Querstriche ragen unter dem Dach hervor
        │     │     │     │     │             (sichtbares Drittel)
   ═════╧═════╧═════╧═════╧═════╧═════     <- durchgehende weisse Längslinie
              F a h r b a h n
```
Die Längslinie ist überall zu sehen, die Querstriche gehen links und rechts jedes Feldes bis zur Dachkante. 5 Querstriche in gleichem Abstand → 6 Felder (bzw. so viele, wie der Abstand zur Normbreite ergibt).

**Wo der Punkt hinkommt, wenn nur ein Teil sichtbar ist.** Der Mittelpunkt des sichtbaren Teils ist **nicht** der Mittelpunkt des Feldes. Wir gehen darum von der sichtbaren Längslinie aus und zeichnen das Kästchen in voller Normtiefe nach hinten, auch unter das Dach:
```
        ┌─────┐  ▲
        │█████│  │  verdeckt (2/3)
        │██●██│  │  <- Feldmitte = Längslinie + halbe Normtiefe (liegt unter dem Dach!)
        │█████│  │
        ├─────┤  │  <- Dachkante
        │     │  │  sichtbar (1/3)
   ═════╧═════╧══▼  <- Längslinie (Anker)
```
Falsch wäre der Punkt in der Mitte des sichtbaren Drittels – das Kästchen läge dann zu weit vorne auf der Fahrbahn.

**Das Markierungsmuster (Vorschlag aus der Besprechung):** Eine weisse Linie entlang der Fahrbahn, davon gehen in regelmässigem Abstand kurze Querstriche ab (siehe Notebook 01, Abbildung `01_schwierige_stellen.png`, Fall E). Kommt dieses Muster **mehrmals nebeneinander** vor, ist das ein starker Beleg für eine Reihe von Parkfeldern – auch wenn Autos die Felder selbst verdecken. Schwierig bleibt das **einzelne Feld** (Fall F): Dort gibt es kein Muster, das sich wiederholt.

### 4.8 Lageabweichung der DAV-Punkte (neu aus Fall H)

Die DAV-Punkte stimmen nicht auf den Meter mit dem Luftbild überein. Für die **Anzahl** je Strassenabschnitt ist das unwichtig: Zwei erkannte Felder und zwei DAV-Punkte im selben Abschnitt ergeben ein Delta von 0. Für das **Lernen** ist es wichtig: Die Parkfeld-Rechtecke (Notebook 04) werden um die DAV-Punkte gezeichnet. Ist der Punkt versetzt, ist auch die Lernmaske versetzt, und das Netz lernt teils Asphalt neben dem Feld als "Parkplatz".

Skizze (Seitenansicht eines Feldes, DAV-Punkt 1 m zu weit links):
```
Luftbild:            |=====  echtes Feld  =====|
Lernmaske heute:  |=====   Rechteck   =====|
                  ^^^                       ^^^
                  Asphalt, als "Parkplatz"  Feld, als "kein Parkplatz" gelernt

```
Beim Lernen vergleicht das Netz seine Antwort Pixel für Pixel mit dem Lösungsblatt; jede Abweichung zählt als Fehler. Ein versetztes Lösungsblatt "bestraft" das Netz also am Rand, obwohl es richtig liegt – es lernt, zu wenig zu melden.

**Mit der eigenen Ground Truth ✅ erledigt (26.11.2026):** Das Lösungsblatt wird aus den von Hand gesetzten Feldern gezeichnet, nicht aus den DAV-Punkten. Die Versetzung der DAV-Punkte stört das Lernen dann nicht mehr; ein unsicherer Rand ist nicht nötig. Für den Vergleich am Schluss spielt die Versetzung keine Rolle, weil nur Anzahlen verglichen werden.

**Nachtrag aus der Sichtprüfung (Notebook 01, Abschnitt 13).** Bei den 20 von Hand beurteilten Feldern lag der DAV-Punkt überall, wo ein Feld zu sehen war, im Feld oder auf einer Linie – nie daneben. Eine Versetzung wie in Fall H kommt also vor, war in dieser Stichprobe aber nicht die Regel.

### 4.10 Parkierte Autos auf dem Feld ✅ (26.11.2026)

**Das Problem.** Auf fast jedem Parkfeld steht im Bild ein Auto. Daraus entstehen drei Fragen: Ist ein Auto eine Verdeckung? Was gilt, wenn Autos alle Linien verdecken? Und lernt das Netz am Ende Autos statt Markierungen?

**Was die Quellen tun – und warum wir es anders machen müssen.** In den Arbeiten, auf die wir uns stützen, ist ein parkiertes Auto ein **Beleg** für Parkfläche: Hellekes et al. (2022) beschriften als Parkfläche, was "amtlich dafür bestimmt ist **oder** zum Zeitpunkt der Aufnahme zum Parkieren benutzt wird"; Henry et al. (2021) schliessen ausdrücklich Flächen ein, die regelmässig beparkt werden, "obwohl keine Markierung sichtbar ist"; bei Rauch et al. (2025) wird eine Strasse nur dann als "auch Parkfläche" beschriftet, wenn im Bild Autos darauf stehen. Hellekes berichtet auch die Kehrseite: ein Lieferwagen in einer Ladezone wurde fälschlich als Parkfläche erkannt. Unser Ziel ist enger – **markierte** Parkplätze (Abschnitt 1). Darum darf bei uns ein Auto weder Beleg noch Hindernis sein. Rauch et al. (2025) halten selbst fest, dass die Abhängigkeit vom Auto nur das Parkieren am Strassenrand **ohne** Markierung betrifft, nicht Flächen, die durch Randstein, Markierung oder Belag abgegrenzt sind.

**Kurz gesagt: Die Markierung entscheidet. Ein Auto ist weder Beleg noch Verdeckung.** Die vier Regeln unten führen diesen einen Satz aus.

**Die vier Regeln ✅ (vom Auftraggeber angenommen, 26.11.2026).**

| Nr. | Regel | Begründung |
| --- | --- | --- |
| 1 | **Ein Auto ist keine Verdeckung.** Beim verdeckten Anteil zählen nur Dinge, unter denen man nicht weiss, was liegt: Baumkrone, Brücke, Dach, Baustelle. | So hat der Auftraggeber die 20 Felder der Sichtprüfung beurteilt. Würde das Auto mitzählen, wäre fast jedes belegte Feld zu mehr als zwei Dritteln "verdeckt" (ein Auto von 1,85 × 4,90 m füllt ein Längsfeld von 1,90 × 5 bis 6 m fast ganz). |
| 2 | **Massgebend ist die Markierung, nicht das Auto.** Ein Feld mit Auto gilt als `sichtbar`, wenn mindestens eine seiner Linien neben oder zwischen den Autos zu erkennen ist und das Muster der Nachbarfelder fortsetzt (Fall 5). Bei einem Feld ohne Nachbarn braucht es zwei Linien, die eine Ecke bilden. | Autos sind schmaler und kürzer als das Feld; bei 5 cm je Pixel ist eine Linie daneben meist zu sehen. |
| 3 | **Verdecken Autos alle Linien eines Feldes, ist das Feld `nicht sichtbar`, verdeckt durch `Fahrzeug`.** Es wird behandelt wie jedes verdeckte Feld: im Lösungsblatt 255, gezählt nur mit Beleg (anderes Luftbild oder amtliche Daten, Abschnitt 9) und getrennt ausgewiesen. | Sonst gäbe es Lernbeispiele, bei denen **nur** Autos für "Parkplatz" sprechen. Genau daraus würde das Netz "Autos in einer Reihe = Parkplatz" lernen. |
| 4 | **Das Kästchen richtet sich nach den Linien, nie nach dem Auto.** | Ein Auto steht nie genau im Feld (Hinweis des Auftraggebers). |

**Was diese Regeln bewirken.** Jedes Lernbeispiel für "Parkplatz" zeigt eine erkennbare Markierung. Fahrende Autos, Autos auf dem Trottoir und falsch parkierte Autos stehen als Gegenbeispiele ("kein Parkplatz") ohnehin in den Lernbildern. Was Parkplatz von Nicht-Parkplatz unterscheidet, ist damit in den Lerndaten die Markierung.

**Was die Regeln nicht verhindern können.** Auch bei einem sichtbaren Feld liegt das Auto im Kästchen und wird als "Parkfläche" gelernt. Das lässt sich nicht vermeiden, solange wir Flächen zählen (Anzahl = Fläche ÷ Platzfläche). Ob das Netz sich trotzdem zu sehr auf Autos stützt, kann man nicht verhindern, aber **messen**:

| Prüfung (nur in den Testzonen) | Was dafür nötig ist | Was sie zeigt |
| --- | --- | --- |
| **Leer gegen belegt**, für 5 cm und 10 cm getrennt: Wie oft findet das Netz ein sichtbares Feld, wenn es leer ist, und wie oft, wenn ein Auto darauf steht? | je Feld der Testzonen ein Häkchen "belegt" oder "leer" | Findet es leere Felder deutlich seltener, stützt es sich auf Autos. Der Vergleich der beiden Auflösungen gehört zur Leitfrage: Eine Parkfeldlinie ist 10 cm breit (Stadt Bern, Normalie 2.10.22), also 2 Pixel bei 5 cm und 1 Pixel bei 10 cm. Ob das Netz bei 10 cm noch die Linie nutzen kann oder sich stärker auf Autos stützt, ist eine offene Frage, die diese Prüfung beantwortet. |
| **Falsche Meldungen ansehen:** eine zufällige Stichprobe der Flächen, die das Netz meldet, wo laut Ground Truth nichts ist | Sichtprüfung von Hand mit festen Wörtern, z. B. `Auto ohne Markierung`, `Markierung übersehen in der Ground Truth`, `anderes` | Wie viele Fehlmeldungen gehen auf Autos zurück? |

Formal: Für sichtbare Felder sei $R_{\text{leer}} = P(\text{gefunden} \mid \text{leer})$ und $R_{\text{belegt}} = P(\text{gefunden} \mid \text{belegt})$, wobei ein Feld als gefunden gilt, wenn mindestens die Hälfte seines Kästchens als Parkfläche gemeldet wird. Ausgewiesen wird $\Delta R = R_{\text{belegt}} - R_{\text{leer}}$ mit einem Bootstrap-Intervall über die Testzonen (7.9). Ein Wert nahe 0 heisst: Das Netz braucht das Auto nicht. Es wird **kein Zielwert** gesetzt; die Zahl wird berichtet und gedeutet.

**Was bewusst nicht dazugehört.** Keine Kunstgriffe beim Lernen (z. B. Autos aus Bildern entfernen, zusätzliche Verlustterme): Sie stehen in keiner der Quellen, und ihr Nutzen wäre hier nicht belegt. Zeigt die Prüfung eine starke Abhängigkeit von Autos, ist das ein Ergebnis dieses Pilots und der Anlass für einen nächsten Schritt.

**Festgelegt ✅:** Das Häkchen "belegt/leer" wird nur in den Testzonen gesetzt. **Noch unbekannt:** Wie oft Regel 3 greift, wissen wir erst mit der Ground Truth; in Strassen, wo nur Ecken markiert sind, könnte es häufig sein. Dann stünden viele Felder unter "verdeckt, belegt" statt unter "erkannt".

### 7.4 Netz und Lernen 🔁

> **Stand heute.** Gilt. Ergänzungen: keine Gewichtung der Klassen (Abschnitt 11, E9); 16 und 40 Epochen (E10); zum Verlängern siehe 11.1 und 11.2 (N2). Der Anteil der Parkfläche in den Kacheln beträgt rund 1,3 % (Notebook 05), nicht unter 1 % wie unten aus Pilot 1 übernommen.

**Verlust.** Kreuzentropie über die bewerteten Pixel:

$$\mathcal{L}(\theta) = -\frac{1}{|\Omega|} \sum_{i \in \Omega} \big[\, y_i \log p_i + (1 - y_i) \log (1 - p_i) \,\big]$$

**Optimierung.** Adam, Lernrate $\eta_e = 10^{-4} \cdot 0{,}937^{\,e}$ in Epoche $e$; 4 Ausschnitte je Schritt, Gradienten über 3 Schritte gesammelt (wirksam 12). Zufällige Drehungen um Vielfache von 90° und Spiegelungen.

**Annahmen und Folgen.**
- *Drehungen und Spiegelungen* setzen voraus, dass Parkfelder keine bevorzugte Himmelsrichtung haben. Für senkrechte Luftbilder trifft das zu; die Schattenrichtung wird dabei mitgedreht, was das Netz gegenüber dem Sonnenstand unempfindlicher macht.
- *Seltenheit.* Der Anteil $\pi$ der Parkflächen-Pixel liegt unter 1 %. In der ungewichteten Kreuzentropie stammen damit über 99 % der Summanden von "keine Parkfläche". Ein Netz, das überall 0 antwortet, hat bereits einen kleinen Verlust; das begünstigt eine tiefe Trefferquote (*Recall*) und damit **Unterzählung**. Mögliche Gegenmittel ❓: Klassengewicht $w_1 > 1$, Focal Loss, Dice-Verlust. Jedes verschiebt den Arbeitspunkt – danach muss die Schwelle $\tau$ (7.5) auf der Validierung neu gewählt werden.
- *Wahl der besten Epoche.* Gespeichert wird $e^* = \arg\max_e \mathrm{IoU}_{\text{val}}(e)$. Das Maximum einer verrauschten Kurve ist nach oben verzerrt, umso mehr, je mehr Epochen verglichen werden. Die Validierungs-IoU bei $e^*$ ist deshalb **kein** Endergebnis; berichtet wird die einmalige Prüfung in den Testzonen.
- *Epochenzahl.* Liegt $e^*$ am Ende des Laufs, ist nicht belegt, dass die Kurve ihr Maximum erreicht hat → Lauf verlängern (Abschnitt 3).

### 7.5 Vorhersage über grosse Flächen 🔁

> **Stand heute.** Die Schwelle ist kein freier Wert mehr: Sie wird je Lauf auf der Validierung gewählt und eingefroren (Abschnitt 11, E9). Gemeldet wird ein Pixel, wenn $\bar d_i > \ln\big(\tau / (1 - \tau)\big)$ (12.7).

Das Netz sieht jeweils $512 \times 512$ Pixel. Grosse Flächen werden mit überlappenden Fenstern abgefahren; für jedes Pixel wird der Unterschied der Rohwerte $d_i = z_{i,1} - z_{i,0}$ über alle Fenster $F_i$ gemittelt, die es enthalten:

$$\bar d_i = \frac{1}{|F_i|} \sum_{f \in F_i} d_i^{(f)}, \qquad \hat y_i = \mathbb{1}\big[\bar d_i > \tau'\big]$$

mit $\tau' = 0$ (entspricht $p_i > 0{,}5$ im einzelnen Fenster).

**Begründung.** Am Fensterrand fehlt dem Netz der Zusammenhang; die Überlappung mittelt diese Randfehler heraus.

**Folge.** $\tau'$ ist ein freier Wert ❓. Er bestimmt das Verhältnis von übersehener zu fälschlich gemeldeter Fläche und wirkt über 7.7 direkt auf die Anzahl.

### 7.6 Von Pixeln zu Flächen 🔁

> **Stand heute.** Entfällt (Abschnitt 11, E7): kein Vereinfachen der Umrisse, keine Mindestfläche, kein Höchstabstand zur Strasse. Damit gibt es an dieser Stelle keine freien Werte mehr.

Zusammenhängende Pixel mit $\hat y_i = 1$ werden zu Polygonen. Drei Filter folgen: Vereinfachung nach Douglas–Peucker mit Toleranz $\varepsilon$ (das vereinfachte Polygon weicht höchstens $\varepsilon$ vom Original ab), Mindestfläche $A_{\min}$, Höchstabstand $d_{\max}$ zum nächsten Strassenstück.

**Folge.** Jeder Filter hat einen Schwellenwert, der die Zählung beeinflusst: $A_{\min}$ entfernt auch echte kleine Bruchstücke (wirkt Richtung Unterzählung), $\varepsilon$ verändert die Fläche in der Grössenordnung $\varepsilon \cdot$ Umfang. Die Werte sind im Lerngebiet festzulegen ❓ und vor der Prüfung einzufrieren.

### 7.7 Von Flächen zu Anzahlen 🔁 / 🆕

> **Stand heute.** Geändert (Abschnitt 11, E7; 12.8): Die Klasse einer gemeldeten Fläche folgt aus ihrer mittleren Tiefe; die Platzfläche ist gemessene Breite × Normtiefe (nicht die Normfläche); die Platz-Äquivalente werden je Ausschnitt summiert und **einmal** gerundet. Die Überlegungen unter "Annahmen und Folgen" gelten weiter.

Für jede Fläche $F$ mit Inhalt $A_F$ und Parkart $k$:

$$\hat n_F = \mathrm{round}\!\left(\frac{A_F}{a_k}\right), \qquad a_k = b_k \cdot t_k, \qquad \hat n_s = \sum_{F \in s} \hat n_F$$

🆕 $a_k$ ist die **Normfläche** der Parkart, nicht mehr ein aus Punktabständen geschätzter Mittelwert.

**Annahmen und Folgen.**
- *Fehlerfortpflanzung.* $\delta \hat n \approx \delta A / a_k$: Ein relativer Flächenfehler geht 1 : 1 in die Anzahl. Die Zählung ist höchstens so gut wie die erkannte Fläche.
- *Rundung je Fläche.* Bruchstücke mit $A_F < a_k / 2$ zählen 0. Zerfällt ein Feld in mehrere Stücke, entsteht systematische Unterzählung. Variante ❓: erst summieren, dann runden, $\hat n_s = \mathrm{round}\big(\sum_{F \in s} A_F / a_k\big)$.
- *Passung von $a_k$ und Lösungsblatt.* Stehen Felder dichter als die Normbreite ($w_{\text{echt}} < b_k$), überlappen die Normrechtecke im Lösungsblatt; je Feld bleibt die Fläche $w_{\text{echt}} \cdot t_k$. Teilt man durch $a_k = b_k \cdot t_k$, entsteht auch bei fehlerfreier Erkennung der Faktor $w_{\text{echt}} / b_k$ (Beispiel Querfelder: $2{,}33 / 2{,}50 = 0{,}93$). $a_k$ muss deshalb der Fläche entsprechen, die ein Feld im Lösungsblatt tatsächlich einnimmt.
- *Parkart.* Eine falsche Parkart $k'$ statt $k$ verfälscht die Anzahl um den Faktor $a_k / a_{k'}$. Solange die Parkart nicht sicher aus dem Bild bestimmt werden kann (Abschnitt 5), ist das eine eigene Fehlerquelle.
- *Flächenzählung setzt voraus,* dass das Netz die **ganze Feldfläche** erkennt – auch leere Felder ohne Auto. Erkennt es vor allem Autos, wird die Fläche und damit die Anzahl zu klein.

### 7.8 Verdeckung: Lückenschluss formal 🆕

**Stand ✅ (26.11.2026).** Der Lückenschluss ist als Konzept festgelegt, wird in diesem Pilot aber **nicht angewendet**, solange $L_{\max}$ nicht belegt ist (siehe "Belegbarkeit" unten und Abschnitt 8, W4).

**Verdeckt** heisst ein Ort, wenn der Pflanzenindex hoch ist (Baumkrone). Schatten allein zählt nicht ✅ (W5):

$$\mathrm{NDVI} = \frac{\mathrm{NIR} - R}{\mathrm{NIR} + R} \in [-1, 1], \qquad V = \{\, u : \mathrm{NDVI}(u) > \tau_N \,\}$$

Entlang einer Strassenseite sei $u$ die Längskoordinate; die erkannten Felder sind Intervalle $I_1, \dots, I_m$. Eine **Lücke** $G = (a, b)$ liegt zwischen zwei aufeinanderfolgenden Intervallen. Sie wird genau dann ergänzt, wenn alle Bedingungen gelten:

1. beide Nachbarn haben dieselbe Parkart $k$;
2. in $G$ liegt kein sichtbares Stück von mindestens einer Feldbreite: Jedes zusammenhängende Teilstück von $G \setminus V$ ist kürzer als $w_k$ ✅ (W3);
3. $|G| \le L_{\max}$.

Zu Bedingung 2: Erst auf einer ganzen Feldbreite müsste eine Markierung zu sehen sein. Ein schmaler Spalt zwischen zwei Kronen belegt weder ein Feld noch sein Fehlen.

Dann: $\hat n_G = \lfloor |G| / w_k \rfloor$ mit $w_k$ = Normlänge eines Feldes entlang der Strasse.

**Teilweise sichtbare Felder** gelten als erkannt, wenn mindestens ein Drittel der Feldtiefe sichtbar ist und sich die Querstriche mit Periode $w_k$ (innerhalb einer Toleranz) über mehrere Felder wiederholen. Das Rechteck wird an der sichtbaren Längslinie verankert und in voller Normtiefe $t_k$ gezeichnet.

**Eigenschaften.**
- Der Schätzer ergänzt nur **zwischen zwei Belegen**. Randstücke, einzelne Felder und vollständig verdeckte Reihen bleiben ungezählt. Der verbleibende Fehler ist deshalb **einseitig**: Es ist eher eine Unterzählung zu erwarten als eine Überzählung. Das ist gewollt (Grundsatz in Abschnitt 1) und muss bei der Deutung des Deltas mitgelesen werden.
- **Abgrenzung zu Hellekes et al. (2022).** Deren Hürdenmodell schätzt $\mathbb{E}[n_s \mid c_s] = P(n_s > 0 \mid c_s) \cdot \mathbb{E}[n_s \mid n_s > 0, c_s]$ aus Merkmalen $c_s$ des Abschnitts (z. B. Strassentyp). Das kann im Mittel über viele Abschnitte erwartungstreu sein, setzt aber Plätze ohne lokalen Beleg und überträgt die Verhältnisse der beobachteten auf die unbeobachteten Strassen. Für eine Zählung "ohne Verfälschung" je Abschnitt ist es darum nicht geeignet.
- Freie Werte ❓: $\tau_N$ und $L_{\max}$. Sie werden im Lerngebiet bestimmt (4.9). Für die Höchstlänge ✅: Sei $q(L)$ der Anteil der verdeckten Lücken mit Länge $\le L$ im Lerngebiet, die laut eigener Ground Truth durchgehend Felder enthalten. Dann $L_{\max} = \max\{L : q(L) \ge q_0\}$ mit $q_0 = 0{,}95$ ✅.

**Begründung von $q_0$ (Entscheidungsregel).** Eine Lücke mit $n$ möglichen Feldern enthalte mit Wahrscheinlichkeit $q$ wirklich Felder. Ergänzen kostet im Fehlerfall $c_{+}$ je erfundenem Feld, Nicht-Ergänzen $c_{-}$ je übersehenem Feld. Erwartete Kosten: Ergänzen $(1-q)\,n\,c_{+}$, Nicht-Ergänzen $q\,n\,c_{-}$. Ergänzen ist genau dann besser, wenn

$$\frac{q}{1-q} > \frac{c_{+}}{c_{-}} \quad\Longleftrightarrow\quad q > q_0 = \frac{c_{+}}{c_{+} + c_{-}}$$

Bei gleichen Kosten wäre $q_0 = 0{,}5$. Der Grundsatz "ohne Verfälschung" heisst $c_{+} \gg c_{-}$: $q_0 = 0{,}95$ entspricht $c_{+}/c_{-} = 19$, $q_0 = 0{,}99$ entspricht 99. Der Wert ist also keine mathematische Konstante, sondern drückt eine Wertung aus.

**Belegbarkeit (Stichprobenumfang).** $q$ ist unbekannt und wird aus $N$ geprüften Lücken geschätzt. Sind alle $N$ richtig, ist die untere einseitige 95-%-Vertrauensgrenze (Clopper–Pearson) $q_{\text{unten}} = 0{,}05^{1/N}$. Für $q_{\text{unten}} \ge q_0$ braucht es

$$N \ge \frac{\ln 0{,}05}{\ln q_0}: \qquad q_0 = 0{,}90 \Rightarrow N \ge 29, \qquad q_0 = 0{,}95 \Rightarrow N \ge 59, \qquad q_0 = 0{,}99 \Rightarrow N \ge 299$$

(mit Fehlern unter den $N$ Lücken entsprechend mehr). **Folge:** Verglichen wird nicht der Schätzwert $\hat q$, sondern die untere Vertrauensgrenze mit $q_0$. Reicht die Zahl der verdeckten Lücken im Lerngebiet nicht aus, ist die Regel nicht belegbar und es wird nicht ergänzt.

### 7.9 Güte- und Vergleichsmasse 🔁 / 🆕

> **Stand heute.** Geändert (12.9): MAE und Bias je Ausschnitt statt je Strassenabschnitt; B1 ist die mittlere Zahl sichtbarer Felder je Lern-Ausschnitt; der Bootstrap zieht ganze Testblöcke; das Delta wird in drei Teile zerlegt, nicht in zwei.

**Pixel-Ebene.** Mit richtig positiven (TP), falsch positiven (FP) und falsch negativen (FN) Pixeln: $\mathrm{IoU} = \frac{TP}{TP + FP + FN}$, Genauigkeit $= \frac{TP}{TP + FP}$, Trefferquote $= \frac{TP}{TP + FN}$.

**Zähl-Ebene.** Über **alle** Strassenabschnitte $s = 1, \dots, S$ der Testzonen (auch solche ohne Parkplatz):

$$\mathrm{MAE} = \frac{1}{S} \sum_{s} |\hat n_s - n_s|, \qquad \mathrm{Bias} = \frac{\sum_s (\hat n_s - n_s)}{\sum_s n_s}$$

Würde man nur die Abschnitte nehmen, die ein Verfahren "anfasst", hinge $S$ vom Verfahren ab und die MAE wären nicht vergleichbar.

**Bezugslinien.** B0: $\hat n_s = 0$. B1: $\hat n_s = \lambda \cdot \ell_s$ mit Abschnittslänge $\ell_s$ und der mittleren Dichte $\lambda$ aus dem Lerngebiet. Ein Modell, das B1 nicht schlägt, hat aus dem Bild nichts Brauchbares gelernt.

**Zerlegung des Deltas** 🆕. Mit der (unbekannten) wahren Anzahl $n^*$:

$$\underbrace{\hat n - n^{\text{DAV}}}_{\text{Delta}} = \underbrace{(\hat n - n^*)}_{\text{Fehler des Modells}} - \underbrace{(n^{\text{DAV}} - n^*)}_{\text{Fehler der amtlichen Daten}}$$

Aus dem Delta allein lassen sich die beiden Teile nicht trennen. Erst die eigene Ground Truth in den Testzonen ($n^{\text{GT}} \approx n^*$) macht beide einzeln messbar – das ist der Grund für die dritte Rolle in Abschnitt 1.

**5 cm gegen 10 cm.** Das 10-cm-Bild entsteht aus dem 5-cm-Bild durch Mittelung von $2 \times 2$ Pixeln. Damit unterscheidet sich **nur** die Abtastung (gleiche Kamera, gleicher Zeitpunkt, gleiche Autos). Folge: Ein echtes 10-cm-Bild hat zusätzlich Unschärfe der Optik; der Versuch zeigt also eher eine **Obergrenze** dessen, was mit 10 cm möglich ist. Verglichen wird gepaart je Abschnitt, $D_s = |\hat n_s^{(10)} - n_s| - |\hat n_s^{(5)} - n_s|$, mit Bootstrap: $B = 10\,000$-mal $S$ Abschnitte mit Zurücklegen ziehen, jeweils $\bar D$ bilden; das 95-%-Intervall sind die 2,5- und 97,5-%-Quantile.

**Annahme und Folge.** Der Bootstrap über Abschnitte behandelt sie als unabhängig. Abschnitte derselben Zone teilen aber Strasse, Schatten und Baumbestand; das Intervall wird dadurch **zu schmal**. Mit mehr Zonen (rund 10) ist ein Bootstrap über ganze Zonen die ehrlichere Variante ❓.

**Versuch A gegen Versuch B.** Die beiden Versuche unterscheiden sich gleichzeitig in Belaubung, Jahr, Sonnenstand, parkierten Autos und Ground Truth. Ein Unterschied im Ergebnis lässt sich deshalb nicht einer einzelnen Ursache (z. B. "Laub") zuschreiben; er beschreibt den Gesamtunterschied der beiden Aufnahmen.

### 7.10 Was das für die Aussagekraft bedeutet

> **Stand heute.** Statt "Testzonen" und "Abschnitte" gelten Testblöcke und Ausschnitte. Die Einschränkung "Intervall bei abhängigen Abschnitten zu schmal" ist entschärft, weil ganze Blöcke gezogen werden. Die vierte Zeile (ergänzte Felder) betrifft den Lückenschluss, der nicht angewendet wird.

### 8.2 Befunde und Entscheide ✅ (26.11.2026)

**W1 – Was wird unter Bäumen als Ground Truth gesetzt?**
*Befund.* Bisher ist nicht festgelegt, ob ein vollständig verdecktes Feld in die Ground Truth kommt. Beides hat einen Haken: Setzt man es als "Parkfläche", lernt das Netz, Parkplätze zu melden, wo nur Laub zu sehen ist – das ist Raten. Lässt man es weg, fehlt es in der wahren Anzahl, und die Höchstlänge (Fall 2) lässt sich gar nicht messen.
*Vorschlag.* Jedes Feld der Ground Truth bekommt einen von drei Zuständen:

| Zustand | Bedingung | im Lösungsblatt | zählt zur wahren Anzahl |
| --- | --- | --- | --- |
| **sichtbar** | mindestens ein Drittel sichtbar (Fall 5) | 1 = Parkfläche | ja |
| **verdeckt, belegt** | im Bild nicht sichtbar, aber anders belegt (anderes Luftbild, amtliche Daten) | 255 = nicht bewertet | ja, getrennt ausgewiesen |
| **unsicher** | z. B. Baustelle | 255 = nicht bewertet | nein |

Damit gibt es zwei wahre Zahlen: *sichtbare Felder* (das kann ein Netz höchstens finden) und *alle Felder* (das wäre vollständig).
*Geprüft.* Logisch gegen den Grundsatz "nicht raten" und gegen 7.8 (die Höchstlänge braucht genau die Felder "verdeckt, belegt").
*Entscheid ✅.* Angenommen: drei Zustände je Feld.

**W2 – Der Blick ins andere Jahr widerspricht dem Versuchsplan.**
*Befund.* Regel 4 in 4.6 ("ist die Stelle im anderen Bild sichtbar, gilt die Erkennung") und der Hinweis in 4.5 ("zuerst das zweite Bild prüfen") greifen auf das Bild eines **anderen Jahres** zu. Abschnitt 3 verlangt aber zwei für sich stimmige Versuche, und Fall D zeigt, warum: Was 2021 ein Parkplatz war, ist 2024/25 eine Baustelle.
*Vorschlag.* In Versuch A und B wird **nur das eigene Bild** verwendet. Das Zusammenlegen beider Jahre ist ein **drittes, getrennt ausgewiesenes Ergebnis** ("Fusion") mit der Einschränkung: Ein Fund aus 2021 darf 2024/25 nur eine nachweislich **verdeckte** Stelle füllen, nie sichtbaren Asphalt und nie eine unsichere Stelle.
*Geprüft.* Gegen Fall D in Notebook 01, Abschnitt 12.
*Entscheid ✅.* Angenommen: je Versuch nur das eigene Bild (nachgeführt in 4.5 und 4.6).

**W3 – "Nachweislich verdeckt" und "sichtbarer Asphalt" überschneiden sich.**
*Befund.* 7.8 verlangt, dass eine Lücke zu einem Anteil $\rho$ verdeckt ist, und zugleich, dass kein sichtbarer Asphalt darin liegt. Ist $\rho$ kleiner als 1, ist der Rest der Lücke per Definition sichtbar und ohne erkanntes Feld – also "sichtbarer Asphalt". Die beiden Bedingungen widersprechen sich.
*Vorschlag.* $\rho$ entfällt. Neue, eine Bedingung: In der Lücke gibt es **kein sichtbares Stück, das mindestens eine Feldbreite lang ist**. Begründung: Erst auf einer ganzen Feldbreite müsste eine Markierung zu sehen sein; ein schmaler Spalt zwischen zwei Kronen beweist nichts.
*Geprüft.* Logisch; ein freier Wert weniger.
*Entscheid ✅.* Angenommen (nachgeführt in 4.6 und 7.8).

**W4 – Die Höchstlänge lässt sich in Kreis 1 voraussichtlich nicht belegen.**
*Befund.* Für "95 von 100" braucht es mindestens 59 fehlerfreie Mittelstück-Lücken. Notebook 01, Abschnitt 12 zeigt: Die Felder unter Baumkronen bilden im ganzen Kreis 1 nur einige Dutzend Gruppen – und darin sind Randstücke und einzelne Felder schon mitgezählt. Mittelstücke gibt es also weniger als Gruppen.
*Folge nach dem geltenden Entscheid.* In diesem Pilot wird **nicht ergänzt**. Stattdessen wird ausgewiesen: "n Felder verdeckt, belegt – vom Netz nicht erkennbar".
*Vorschlag.* So belassen. Der Lückenschluss bleibt als geprüftes Konzept im Leitfaden und kann später belegt werden, wenn Lücken aus weiteren Kreisen dazukommen (dafür müssten nur die Lücken beurteilt werden, nicht ganze Gebiete).
*Geprüft.* An den Daten (Notebook 01, Abschnitt 12) und an der Rechnung in `formeln_pruefen`, Prüfung 5.
*Entscheid ✅.* Angenommen: In diesem Pilot wird nicht ergänzt, solange die Höchstlänge nicht belegt ist. Die Zahl der Mittelstück-Lücken wird mit der eigenen Ground Truth trotzdem gezählt und ausgewiesen; sie zeigt, wie weit wir von den nötigen 59 entfernt sind.

**W5 – Schatten ist nicht dasselbe wie eine Baumkrone.**
*Befund.* 4.2 behandelt "sehr dunkel" wie "unter einer Krone" (beides `B`). Fall D zeigt aber, dass Felder im Schatten oft erkennbar bleiben.
*Vorschlag.* Als "verdeckt" gilt vorerst **nur die Baumkrone** (NDVI). Felder im Schatten gelten als sichtbar, wenn die Beschriftungsregel (ein Drittel) erfüllt ist; das Netz muss sie lernen.
*Geprüft.* Gegen die Fälle D, F, G in Notebook 01.
*Entscheid ✅.* Angenommen (nachgeführt in 4.2, 4.5 und 7.8).

**W6 – Parkstreifen ohne Einzelmarkierung.**
*Befund.* Das Ziel heisst "alle markierten öffentlichen Parkplätze". In Kreis 1 gibt es neben den Einzelfeldern auch Parkstreifen als Linien (Notebook 01, Abschnitt 8: Güterumschlag, Güter/Taxi, Blaue Zone). Dort sind keine einzelnen Felder aufgemalt; eine Anzahl gibt es nur als Länge ÷ Platzlänge.
*Frage.* Zählen Streifen mit?
*Entscheid ✅.* **Nein.** Wo nichts aufgemalt ist, kann das Netz nichts erkennen; solche Streifen zählen nicht mit. Wo Felder aufgemalt sind, gehören sie zum Ziel – egal ob Güterumschlag, Taxi oder Blaue Zone. Das entspricht dem Ziel aus Abschnitt 1 ("alle **markierten** öffentlichen Parkplätze"): Massgebend ist die Markierung im Bild, nicht die Kategorie in den amtlichen Daten.

| Was ist im Bild aufgemalt? | im Lösungsblatt | zählt |

| einzelne Felder | 1 = Parkfläche | ja |
| gar nichts | 0 = keine Parkfläche | nein |
| nur der Rand des Streifens, keine einzelnen Felder | 255 = nicht bewertet | nein; die Länge in Metern wird getrennt ausgewiesen |

*Geprüft.* Gegen das Ziel in Abschnitt 1 und den Grundsatz "nicht raten".

## 9. Sichtprüfung von Hand: einheitliches Schema ✅ (26.11.2026)

**Worum es geht.** Der Auftraggeber beurteilt einzelne Felder von Auge auf den Luftbildern (zuerst die 20 Felder aus Notebook 01, Abschnitt 13). Damit jede Beurteilung gleich lautet und sich auszählen lässt, gibt es für jede Angabe eine **feste Liste von Wörtern**. Freie Worte stehen nur in der Spalte `bemerkung`.

**Abgrenzung.** Die Sichtprüfung ist **nicht** das Lösungsblatt für das Lernen. Das Lösungsblatt entsteht später aus Punkten, die auf dem Luftbild gesetzt werden. Die Sichtprüfung benutzt aber dieselben Wörter für den Zustand eines Feldes (Abschnitt 8.2, W1) – so passen beide zusammen.

**Ablage:** `daten/ground_truth/sichtpruefung_nach_befliegung.csv`, eine Zeile je Feld und Aufnahme.

| Spalte | erlaubte Werte | Bedeutung |
| --- | --- | --- |
| `nr` | 1 bis 20 | Nummer aus Notebook 01, Abschnitt 13 |
| `dav_id` | Kennung der DAV | damit das Feld eindeutig wiedergefunden wird |
| `aufnahme` | `Frühjahr 2021`, `Sommer 2024/25`, `Sommer 2026` | welches Bild beurteilt wurde |
| `form` | `Einzelfeld`, `gelbes Feld mit Kreuz`, `Streifen ohne Einzelfelder`, `unklar` | Was ist aufgemalt? `Einzelfeld` = Linien trennen einzelne Plätze ab. `gelbes Feld mit Kreuz` = gelbes Rechteck mit diagonalem Kreuz von Ecke zu Ecke (Beispiel: Feld Nr. 9); amtlich ein Parkverbotsfeld (SSV, Art. 79a); es wird auf Stufe 0 mitgezählt und ab Stufe 1 getrennt ausgewiesen (Abschnitt 1). `Streifen ohne Einzelfelder` = nur eine Linie mit Abschluss an beiden Enden, ohne Unterteilung (Beispiel: Feld Nr. 7, gelbe Linie). Ein Streifen wird nach 8.2 W6 **nicht gezählt**; seine Länge wird getrennt ausgewiesen |
| `markierung` | `ganz sichtbar`, `teilweise sichtbar`, `nicht sichtbar`, `kein Parkfeld`, `unklar` | Was ist von der aufgemalten Markierung zu sehen? `kein Parkfeld` = die Stelle ist frei einsehbar, aber es ist nichts aufgemalt |
| `verdeckt_anteil` | `0`, `unter 1/3`, `1/3 bis 2/3`, `über 2/3`, `ganz`, `nicht geschätzt` | geschätzter Anteil des Feldes, der verdeckt ist |
| `verdeckt_durch` | `nichts`, `Äste ohne Laub`, `Baumkrone`, `Brücke`, `Brücke und Baumkrone`, `Baustelle`, `Baustelle und Baumkrone`, `anderes` | was das Feld verdeckt. Ein parkiertes Auto zählt **nicht** dazu (so wurden alle 20 Felder beurteilt; Vorschlag für die Regel in 4.10) |
| `schatten` | `kein`, `leicht`, `stark`, `nicht angegeben` | Liegt Schatten auf dem Feld? Schatten ist **keine** Verdeckung (8.2 W5) und steht darum in einer eigenen Spalte |

| `zustand` | `sichtbar`, `verdeckt, belegt`, `verdeckt`, `unsicher`, `kein Parkfeld` | wird aus den Spalten davor **abgeleitet** (Regeln unten) |
| `beleg` | `nicht nötig`, `anderes Luftbild`, `amtliche Daten`, `keiner` | nur für verdeckte Felder: Woher wissen wir, dass dort ein Feld ist? (Regel unten) |
| `punktlage` | `im Feld`, `auf der Linie`, `ausserhalb`, `kein Feld sichtbar` | Wo liegt der DAV-Punkt im Verhältnis zum sichtbaren Feld? `ausserhalb` = ein Feld ist sichtbar, der Punkt liegt daneben (geschätzter Abstand in der Bemerkung). `kein Feld sichtbar` = an der Stelle des Punktes ist im Bild kein Parkfeld zu sehen (Baustelle, Baumkrone, Brücke); die Lage lässt sich darum nicht angeben. Ob der Punkt mittig oder am Rand liegt, steht in der Bemerkung |
| `bemerkung` | freier Text | alles Weitere, möglichst in den Worten des Auftraggebers |

**Regel für fehlende Angaben ✅ (26.11.2026, gilt für alle 20 Felder).** Nennt der Auftraggeber ein Merkmal nicht, heisst das: Es ist nicht vorhanden (kein Schatten, nichts verdeckt). `nicht angegeben` steht nur dort, wo ein Schatten vorhanden ist, aber sein Anteil nicht genannt wurde.

**Regel für `zustand`** (folgt aus den Entscheiden in 4.4 Fall 5, 4.5 und 8.2 W1):

| wenn … | dann `zustand` |
| --- | --- |
| `markierung` ist `ganz sichtbar` oder `teilweise sichtbar` (Linie und Querstriche erkennbar) **und** höchstens zwei Drittel sind verdeckt | `sichtbar` |
| verdeckt durch `Baumkrone`, `Brücke` oder beides (keine Baustelle beteiligt), und `markierung` ist `nicht sichtbar` oder mehr als zwei Drittel sind verdeckt (weniger als ein Drittel sichtbar, Fall 5 nicht erfüllt; Beispiel: Feld Nr. 8 im Bild 2026, etwa 90 % verdeckt) | `verdeckt, belegt`, wenn es einen Beleg gibt (nächste Regel); sonst `verdeckt` |
| `markierung` ist `nicht sichtbar` und eine Baustelle ist beteiligt, oder `markierung` ist `unklar`, oder `form` ist `unklar`, obwohl mindestens ein Drittel sichtbar ist (man sieht genug, kann aber nicht sagen, was aufgemalt ist) | `unsicher` |
| `markierung` ist `kein Parkfeld` | `kein Parkfeld` |

**Regel für `beleg` ✅ (26.11.2026).** Ein verdecktes Feld sieht man im Bild nicht. Dass dort trotzdem eines ist, braucht einen Beleg von aussen:

| Lage | `beleg` | Beispiel |
| --- | --- | --- |
| Das Feld ist in einem **anderen Luftbild** sichtbar | `anderes Luftbild` | Nr. 6: im Sommer unter der Baumkrone, im Frühjahr 2021 klar zu sehen |
| Das Feld ist in **keinem** Luftbild sichtbar, steht aber in den **amtlichen Daten** | `amtliche Daten` | Nr. 17: in allen drei Bildern unter der Baumkrone, in der DAV eingetragen |
| weder noch | `keiner` | das Feld bleibt `verdeckt` und zählt nicht |

Drei Dinge gehören dazu:
- **Nachbarfelder allein sind kein Beleg.** Aus "links und rechts sind Felder" auf ein Feld dazwischen zu schliessen, wäre ein Schluss aus dem Zusammenhang (Grundsatz in Abschnitt 1).
- **Kein Widerspruch zu Abschnitt 8.2, W2.** Dort steht: Das *Netz* schaut in jedem Versuch nur das eigene Bild an. Hier geht es um die *Ground Truth*, also um die Lösung, an der das Netz gemessen wird. Sie darf alles nutzen, was wir wissen. Verdeckte, belegte Felder werden zudem immer **getrennt** ausgewiesen (W1): Sie zeigen, was dem Netz gar nicht möglich war.
- **Ein anderes Jahr ist ein schwächerer Beleg als dasselbe Jahr.** Markierungen ändern sich: In der Sichtprüfung war bei 4 von 20 Feldern zwischen den Aufnahmen etwas anderes aufgemalt (Nr. 8, 9, 19, 20). Darum steht in `beleg`, woher der Beleg stammt, und eine **Baustelle** macht eine Stelle immer `unsicher`, auch wenn ein anderes Bild dort ein Feld zeigt.

**Ergänzung zu Fall 5 ✅ (aus der Sichtprüfung, Feld Nr. 4).** Womit ein Feld teilweise verdeckt ist, spielt keine Rolle: Dach, Baumkrone oder Baustellenmaterial werden gleich behandelt. Sind die Linie entlang der Fahrbahn und die Querstriche erkennbar und ist mindestens ein Drittel sichtbar, gilt das Feld als `sichtbar`. Sind die Linien dagegen vollständig bedeckt (Felder Nr. 3 im Bild 2026 und Nr. 5 im Bild 2024/25), gilt die Stelle als `unsicher`.

**Hinweis.** Parkierte Autos sind ein Hinweis auf ein Parkfeld, aber kein Beleg. Massgebend ist die Markierung (Ziel in Abschnitt 1).

### 12.8 Zählung (Notebook 09; Entscheide E6 und E7)

- **Was gezählt wird:** nur gemeldete Fläche auf bewerteten Pixeln ($y_i \in \{0, 1\}$). Was das Netz auf verdeckten Stellen und auf Streifen meldet, wird getrennt ausgewiesen. Meldungen an unsicheren Stellen und im Kreis um einzeln stehende Felder zählen nicht und werden in der Auswertung nicht ausgewiesen; in der Karte gehören sie zur Gruppe "gemeldet, nicht gezählt" (11.3, K4).
- **Mittlere Tiefe:** Für jede zusammenhängende gemeldete Fläche $F$ mit Inhalt $A_F$ sei $\ell_F$ die lange Seite des kleinsten umschliessenden Rechtecks. Dann $t_F = A_F / \ell_F$.
- **Klasse der Fläche:** *eng*, wenn $t_F > 3{,}45$ m (die Mitte zwischen 1,90 m und 5,00 m), sonst *weit*.
- **Platz-Äquivalente:** $E_F = A_F / a_{k(F)}$.
- **Anzahl je Ausschnitt:** $\hat n_u = \big\lfloor \sum_{F \in u} E_F + 0{,}5 \big\rfloor$. Gerundet wird einmal je Ausschnitt, kaufmännisch. Es gibt keinen Filter für kleine Flächen und kein Glätten.
- **Vergleichsraum:** Eine gemeldete Fläche zählt dazu, wenn ihr Mittelpunkt im Strassenraum und im Gebiet liegt.
- **Grenze der Formel:** Wird ein Querfeld allein gemeldet, ist seine lange Seite die Tiefe; die Formel gibt ihm die Klasse *weit*. Es zählt dann 1,17 statt 1,00 Platz-Äquivalente (Prüfung 15).
- **Nachgerechnet:** Prüfungen 7 und 13 (Rundung), 15 (mittlere Tiefe und Klasse), 16 (Platz-Äquivalente, einmal runden).

### 12.9 Auswertung (Notebook 10)

- **Fläche**, über alle bewerteten Pixel der Testausschnitte: $\mathrm{IoU} = \frac{TP}{TP + FP + FN}$, Genauigkeit $= \frac{TP}{TP + FP}$, Trefferquote $= \frac{TP}{TP + FN}$, Flächenverhältnis $= \frac{TP + FP}{TP + FN}$.
- **Anzahl**, über alle $U$ Testausschnitte, auch die leeren:

$$\mathrm{MAE} = \frac{1}{U} \sum_{u} |\hat n_u - n_u|, \qquad \mathrm{Bias} = \frac{\sum_u \hat n_u}{\sum_u n_u} - 1$$

- **Dazu:** dieselbe MAE nur über Ausschnitte mit $n_u > 0$, und als Untergrenze die MAE der Zählformel auf dem Lösungsblatt.
- **5 cm gegen 10 cm:** $D_u = |\hat n_u^{(10)} - n_u| - |\hat n_u^{(5)} - n_u|$; $\bar D$ ist das Mittel über alle Testausschnitte.
- **Unsicherheit:** Bootstrap über **ganze Testblöcke**. 10'000-mal werden so viele Blöcke gezogen, wie es Testblöcke gibt, mit Zurücklegen (Seed 42). Für jede Ziehung ist $\bar D^*$ die Summe von $D_u$ über alle Ausschnitte der gezogenen Blöcke, geteilt durch die Zahl dieser Ausschnitte. Das 95-%-Intervall reicht vom 2,5-%- zum 97,5-%-Quantil.
- **Delta im Vergleichsraum.** Mit $M$ = gezählte Plätze der Methode, $S$ = sichtbare Felder, $V$ = verdeckt, belegt, $Z$ = einzeln stehend, $A$ = amtliche Zahl:

$$\underbrace{M - A}_{\text{Delta}} = \underbrace{(M - S)}_{\text{1. Fehler der Methode}} \;+\; \underbrace{-(V + Z)}_{\text{2. nicht sichtbar}} \;+\; \underbrace{(S + V + Z - A)}_{\text{3. Ground Truth gegen amtlich}}$$

- **Unsicherheit des Deltas in Prozent:** derselbe Bootstrap über Blöcke mit $\sum M^* / \sum A^* - 1$.
- **Sonderfelder (11.2, N1):** Die Trefferquote wird getrennt gerechnet für Pixel in Kästchen sichtbarer Sonderfelder (`✕`, `?`) und für die übrigen Pixel sichtbarer Felder.
- **Nachgerechnet:** Prüfungen 1, 2 (MAE und Bias), 17 (Bootstrap über Blöcke), 18 (Zerlegung des Deltas).
